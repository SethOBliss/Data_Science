### Objective
Forecast daily USD/INR exchange rates using classical time series models (ARIMA and SARIMA), evaluate their performance on held-out test data using MAE, and compare which model better captures the dynamics of the currency pair.

## 1. Import Libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
import warnings
warnings.filterwarnings('ignore')

import yfinance as yf

from statsmodels.tsa.stattools import adfuller, acf, pacf
from statsmodels.graphics.tsaplots import plot_acf, plot_pacf
from statsmodels.tsa.arima.model import ARIMA
from statsmodels.tsa.statespace.sarimax import SARIMAX
from statsmodels.stats.diagnostic import acorr_ljungbox

from sklearn.metrics import mean_absolute_error

plt.rcParams['figure.figsize'] = (12, 5)
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.3

print('All libraries loaded successfully.')
```

---
## 2. Data Collection

We use `yfinance` to pull daily USD/INR closing prices. The ticker `INR=X` on Yahoo Finance represents the US Dollar expressed in Indian Rupees.

```python
# Download USD/INR daily closing prices
ticker = 'INR=X'
start_date = '1995-01-01'
end_date   = '2023-09-30'

raw = yf.download(ticker, start=start_date, end=end_date, auto_adjust=True)
df = raw[['Close']].copy()
df.columns = ['USD_INR']
df.dropna(inplace=True)

print(f'Dataset shape: {df.shape}')
print(f'Date range:    {df.index.min().date()} to {df.index.max().date()}')
df.tail()
```

---
## 3. Exploratory Data Analysis

Before modelling, we visually inspect the series for trends, volatility clustering, and potential seasonality.

```python
fig, axes = plt.subplots(3, 1, figsize=(13, 10), sharex=True)

# Raw series
axes[0].plot(df.index, df['USD_INR'], color='steelblue', linewidth=0.9)
axes[0].set_title('USD/INR Daily Closing Price', fontsize=13)
axes[0].set_ylabel('INR per 1 USD')

# Rolling mean & std (30-day window)
roll_mean = df['USD_INR'].rolling(window=30).mean()
roll_std  = df['USD_INR'].rolling(window=30).std()

axes[1].plot(df.index, df['USD_INR'], color='steelblue', linewidth=0.7, label='Original', alpha=0.6)
axes[1].plot(df.index, roll_mean,  color='firebrick',  linewidth=1.5, label='30-day Rolling Mean')
axes[1].set_title('Rolling Mean (30-day)', fontsize=13)
axes[1].set_ylabel('INR per 1 USD')
axes[1].legend()

axes[2].plot(df.index, roll_std, color='darkorange', linewidth=1.2)
axes[2].set_title('Rolling Std Deviation (30-day)', fontsize=13)
axes[2].set_ylabel('Std Dev')

plt.tight_layout()
plt.savefig('eda_plot.png', dpi=150, bbox_inches='tight')
plt.show()
```

**Observation:** The rolling mean shows a clear upward trend (INR depreciation over time), and the rolling std is not flat — both indicators suggest the series is **non-stationary**. We confirm this formally with the ADF test below.

---
## 4. Stationarity Testing — Augmented Dickey-Fuller Test

The ADF test checks the null hypothesis that a unit root is present (i.e., the series is non-stationary).  
- **Fail to reject H₀** (p > 0.05): series is non-stationary → differencing required  
- **Reject H₀** (p ≤ 0.05): series is stationary → ready for modelling

```python
def adf_test(series, label='Series'):
    result = adfuller(series.dropna(), autolag='AIC')
    print(f'--- ADF Test: {label} ---')
    print(f'  ADF Statistic : {result[0]:.4f}')
    print(f'  p-value       : {result[1]:.4f}')
    for key, val in result[4].items():
        print(f'  Critical ({key}): {val:.4f}')
    conclusion = 'STATIONARY ✓' if result[1] <= 0.05 else 'NON-STATIONARY ✗ — differencing needed'
    print(f'  Conclusion    : {conclusion}\n')
    return result[1]

# Test on original series
p_orig = adf_test(df['USD_INR'], label='Original USD/INR')

# First-order differencing
df['USD_INR_diff1'] = df['USD_INR'].diff()

# Test on differenced series
p_diff = adf_test(df['USD_INR_diff1'].dropna(), label='First-Differenced USD/INR')
```

```python
# Plot differenced series
plt.figure(figsize=(12, 4))
plt.plot(df.index, df['USD_INR_diff1'], color='steelblue', linewidth=0.8)
plt.axhline(0, color='firebrick', linewidth=1, linestyle='--')
plt.title('First-Differenced USD/INR (d=1)', fontsize=13)
plt.ylabel('Δ INR per 1 USD')
plt.tight_layout()
plt.savefig('differenced_series.png', dpi=150, bbox_inches='tight')
plt.show()
```

---
## 5. ACF & PACF Analysis

We use ACF (Autocorrelation Function) and PACF (Partial ACF) plots on the stationary (differenced) series to identify candidate AR (p) and MA (q) orders for ARIMA:
- **PACF cuts off at lag p** → AR(p) structure
- **ACF cuts off at lag q** → MA(q) structure

```python
stationary_series = df['USD_INR_diff1'].dropna()

fig, axes = plt.subplots(1, 2, figsize=(14, 4))

plot_acf(stationary_series,  lags=40, ax=axes[0], title='ACF — Differenced Series',  color='steelblue')
plot_pacf(stationary_series, lags=40, ax=axes[1], title='PACF — Differenced Series', color='firebrick')

plt.tight_layout()
plt.savefig('acf_pacf.png', dpi=150, bbox_inches='tight')
plt.show()
```

---
## 6. Train / Test Split

```python
split_ratio = 0.80
n = len(df)
split_idx = int(n * split_ratio)

train = df['USD_INR'].iloc[:split_idx]
test  = df['USD_INR'].iloc[split_idx:]

print(f'Train size : {len(train)} observations ({train.index.min().date()} → {train.index.max().date()})')
print(f'Test size  : {len(test)}  observations ({test.index.min().date()} → {test.index.max().date()})')
```

---
## 7. ARIMA Model

Based on ACF/PACF analysis, we set d=1 (confirmed by ADF test) and select p, q from the plots above.

```python
# --- Set your identified parameters here ---
p, d, q = 1, 1, 1   # Update after inspecting ACF/PACF

arima_model = ARIMA(train, order=(p, d, q))
arima_fit   = arima_model.fit()

print(arima_fit.summary())
```

```python
# Forecast on test set
arima_forecast = arima_fit.forecast(steps=len(test))
arima_mae      = mean_absolute_error(test, arima_forecast)

print(f'ARIMA({p},{d},{q}) — Test MAE: {arima_mae:.4f} INR')

# Plot
plt.figure(figsize=(13, 5))
plt.plot(train.index, train,          color='steelblue',  linewidth=0.8, label='Train')
plt.plot(test.index,  test,           color='darkorange', linewidth=1.2, label='Actual (Test)')
plt.plot(test.index,  arima_forecast, color='firebrick',  linewidth=1.2, linestyle='--', label=f'ARIMA Forecast (MAE={arima_mae:.3f})')
plt.title(f'ARIMA({p},{d},{q}) — USD/INR Forecast vs Actual', fontsize=13)
plt.ylabel('INR per 1 USD')
plt.legend()
plt.tight_layout()
plt.savefig('arima_forecast.png', dpi=150, bbox_inches='tight')
plt.show()
```

---
## 8. SARIMA Model

SARIMA extends ARIMA with seasonal components (P, D, Q, s), useful if the currency exhibits weekly or monthly cycles.  
We set s=5 (5 trading days per week) as an initial assumption.

```python
# --- Set your identified parameters here ---
P, D, Q, s = 1, 1, 1, 5   # Update based on seasonal ACF/PACF

sarima_model = SARIMAX(train, order=(p, d, q), seasonal_order=(P, D, Q, s))
sarima_fit   = sarima_model.fit(disp=False)

print(sarima_fit.summary())
```

```python
# Forecast on test set
sarima_forecast = sarima_fit.forecast(steps=len(test))
sarima_mae      = mean_absolute_error(test, sarima_forecast)

print(f'SARIMA({p},{d},{q})({P},{D},{Q},{s}) — Test MAE: {sarima_mae:.4f} INR')

# Plot
plt.figure(figsize=(13, 5))
plt.plot(train.index, train,           color='steelblue',  linewidth=0.8,  label='Train')
plt.plot(test.index,  test,            color='darkorange', linewidth=1.2,  label='Actual (Test)')
plt.plot(test.index,  sarima_forecast, color='purple',     linewidth=1.2,  linestyle='--', label=f'SARIMA Forecast (MAE={sarima_mae:.3f})')
plt.title(f'SARIMA({p},{d},{q})({P},{D},{Q},{s}) — USD/INR Forecast vs Actual', fontsize=13)
plt.ylabel('INR per 1 USD')
plt.legend()
plt.tight_layout()
plt.savefig('sarima_forecast.png', dpi=150, bbox_inches='tight')
plt.show()
```

---
## 9. Model Comparison

```python
results = pd.DataFrame({
    'Model': [f'ARIMA({p},{d},{q})', f'SARIMA({p},{d},{q})({P},{D},{Q},{s})'],
    'AIC'  : [arima_fit.aic,   sarima_fit.aic],
    'BIC'  : [arima_fit.bic,   sarima_fit.bic],
    'MAE'  : [arima_mae,       sarima_mae]
})

print(results.to_string(index=False))

best = results.loc[results['MAE'].idxmin(), 'Model']
print(f'\nBest model by MAE: {best}')
```

```python
# Side-by-side forecast comparison
plt.figure(figsize=(13, 5))
plt.plot(test.index, test,            color='darkorange', linewidth=1.5,  label='Actual')
plt.plot(test.index, arima_forecast,  color='firebrick',  linewidth=1.2,  linestyle='--', label=f'ARIMA  (MAE={arima_mae:.3f})')
plt.plot(test.index, sarima_forecast, color='purple',     linewidth=1.2,  linestyle=':',  label=f'SARIMA (MAE={sarima_mae:.3f})')
plt.title('ARIMA vs SARIMA — Forecast Comparison (Test Period)', fontsize=13)
plt.ylabel('INR per 1 USD')
plt.legend()
plt.tight_layout()
plt.savefig('model_comparison.png', dpi=150, bbox_inches='tight')
plt.show()
```

---
## 10. Residual Diagnostics

A well-fitted model should have residuals that resemble white noise — no autocorrelation, roughly normally distributed.

```python
# Use the better model's residuals (update 'fit' to whichever performed better)
best_fit = arima_fit   # Change to sarima_fit if SARIMA won
residuals = best_fit.resid

fig, axes = plt.subplots(1, 3, figsize=(15, 4))

# Residual plot
axes[0].plot(residuals, color='steelblue', linewidth=0.8)
axes[0].axhline(0, color='firebrick', linestyle='--')
axes[0].set_title('Residuals Over Time')

# Histogram
axes[1].hist(residuals, bins=40, color='steelblue', edgecolor='white')
axes[1].set_title('Residual Distribution')

# ACF of residuals
plot_acf(residuals, lags=30, ax=axes[2], color='steelblue', title='ACF of Residuals')

plt.tight_layout()
plt.savefig('residual_diagnostics.png', dpi=150, bbox_inches='tight')
plt.show()

# Ljung-Box test for residual autocorrelation
lb_test = acorr_ljungbox(residuals, lags=[10, 20], return_df=True)
print('Ljung-Box Test (H₀: no autocorrelation in residuals):')
print(lb_test)
```

The strong seasonal component suggests that day-of-week effects in trading volume and liquidity play a meaningful role in USD/INR dynamics over a 20-year horizon.
